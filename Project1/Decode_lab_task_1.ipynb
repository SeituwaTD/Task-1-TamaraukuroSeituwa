```python
#Checking for the types of values in the column with missing values and the
# number of unique values
df['CouponCode'].unique()
df['CouponCode'].nunique()
```

```python
#Standardising the column names by turning all to lower case and removing whitespaces
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
)
```

```python
# the column couponcode has 25.75% of empyt rows
missing = df.isnull().mean()*100
print(missing[missing>0].sort_values(ascending=False))
```

```python
df.head()
```

```python
#Showed the data sets had no duplicates
n_dupes = df.duplicated().sum()
print(n_dupes)
```

```python
#fill empty rows of the couponcode column with unkown
df['couponcode'] = df['couponcode'].fillna('unknown')
df['couponcode'].unique()
```

```python
#Checking to see if the missing rows have been removed
missing = df.isnull().mean()*100
print(missing[missing>0].sort_values(ascending=False))
```

```python
print(missing)
```

```python
#Converting every string on the data frame to lower case and removing whitespaces
df = df.map(lambda x:x.strip().lower() if isinstance(x, str) else x)
df.head()
```

```python
#Changing the date column from object to date time
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')
df.head()
```

```python
#Checking to see if the change worked
df.dtypes
```

```python
#Rounding up all float columns to 2 decimal places
float_cols = df.select_dtypes(include='float').columns
df[float_cols] = df[float_cols].round(2)
df.head()
```
